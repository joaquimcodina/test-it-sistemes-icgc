```python
#Importar llibreria pandas
import pandas as pd
```

```python
#Importar fitxers CSV (products.csv i orders.csv)
products = pd.read_csv('products.csv')
orders = pd.read_csv('orders.csv')
```

```python
#Crear nou DataFrame 'order_prices' amb les columnes 'id' i 'euros'.
order_prices = pd.DataFrame(columns=["id", "euros"])

#Afegir dades 'id' del DataFrame 'orders' a 'order_prices'
#NOTA: 'euros' encara no te cap valor.
order_prices["id"] = orders['id']
order_prices
```

```python
#Tipus de dades del DataFrame 'order_prices'
order_prices.dtypes
```

```python
#Realitzar un split de la columna "products" del DataFrame "orders", per separar els IDs dels productes.
#Fem un DataFrame transposat (50x10 -> 10x50) amb l'opció T per facilitar el merge amb 'products'.
df_transposed = orders['products'].str.split(expand=True).T
df_transposed
```

```python
#Bucle for per recorrer el DataFrame transposat per columnes
for col in df_transposed:
    #Crear un aux_col amb la columna actual del DataFrame transposat
    aux_col = pd.DataFrame({"id": df_transposed[col]})

    #Convertir els valors de la columna 'aux_col' de string a float64
    aux_col = aux_col.astype('float64')

    #Realitzar un merge amb la columna 'aux_col' del DataFrame transposat amb 'products'
    #La idea es aconseguir el cost del producte que coincideixin amb el ID (on='id').
    #La opció how='inner' ens permet relacionar les columnes d'ambdos DataFrames amb el mateix ID (evitant nuls com faria 'left' i 'right').
    aux_col = pd.merge(aux_col, products, how='inner', on='id')

    #Sumar els costos de la comanda amb sum() i afegir el cost toal en la columna 'euros'.
    order_prices.loc[col, 'euros'] = aux_col['cost'].sum()

order_prices
```

```python
#Exportar el DataFrame en un fitxer CSV (order_prices.csv)
order_prices.to_csv("order_prices.csv", index=False)
```
