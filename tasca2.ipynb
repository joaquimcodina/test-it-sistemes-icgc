```python
#Importar llibreria pandas
import pandas as pd
```

```python
#Importar fitxers CSV (products.csv i orders.csv)
products = pd.read_csv('products.csv')
orders = pd.read_csv('orders.csv')
```

```python
#Crear nou DataFrame 'product_customers' amb les columnes 'id' i 'customer_ids'.
product_customers = pd.DataFrame(columns=["id", "customer_ids"])

#Copiar dades 'id' del DataFrame 'products' a 'customer_ids'
#NOTA: 'customer_ids' encara no te cap valor.
product_customers["id"] = products['id']
product_customers
```

```python
#Tipus de dades del DataFrame 'product_customers'
product_customers.dtypes
```

```python
#Fer un split en la columna 'products' del DataFrame 'orders', generant un list dels IDs separats per espais.
orders['products'] = orders['products'].str.split()
orders
```

```python
#For loop en el DataFrame 'product_customers'
for index_product_customer, row_product_customer in product_customers.iterrows():    
    customers_list = [] #list on guardarem els IDs dels clients interessats en el producte actual.

    #For loop en el DataFrame 'orders'
    for index_orders, row_orders in orders.iterrows():
        #Comprovar que el ID del producte esta en el llistat de productes de 'orders'
        if product_customers['id'][index_product_customer].astype('str') in orders['products'][index_orders]:
            #En cas afirmatiu, guardar ID de 'customer' que ha comprat aquell producte
            customers_list.append(orders['customer'][index_orders])

    #Guardar els IDs de 'customers_list' que han comprat el producte en la columna 'customer_ids' del DataFrame.
    #Nota: he utilitzat set() per evitar IDs repetits en un mateix producte.
    product_customers.at[index_product_customer, 'customer_ids'] = list(set(customers_list))

product_customers
```

```python
#Exportar el DataFrame en un fitxer CSV (product_customers.csv)
product_customers.to_csv("product_customers.csv", index=False)
```
