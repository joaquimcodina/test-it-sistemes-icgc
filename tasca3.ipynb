```python
#Importar fitxers CSV (customers.csv, orders.csv i order_prices.csv)
customers = pd.read_csv('customers.csv')
orders = pd.read_csv('orders.csv')
order_prices = pd.read_csv('order_prices.csv')
```

```python
#Realitzar un merge dels DataFrames 'order_prices' i 'orders'.
#La idea es aconseguir les columnes de 'orders' coincidint amb el ID (on='id').
#La opció how='inner' ens permet relacionar les columnes d'ambdos DataFrames amb el mateix ID (evitant nuls com faria 'left' i 'right').
df_merged = pd.merge(order_prices, orders, how='inner', on='id')
df_merged
```

```python
#Tipus de dades del DataFrame merged 'df_merged'
df_merged.dtypes
```

```python
#Crear nou DataFrame 'customer_ranking' amb les columnes 'id', 'firstname', 'lastname', 'total_euros'.
customer_ranking = pd.DataFrame(columns=["id", "firstname", "lastname", "total_euros"])

#Copiar dades 'id', 'firstname' i 'lastname' del DataFrame 'customers' a 'customer_ranking'
#NOTA: 'total_euros' encara no te cap valor.
customer_ranking["id"] = customers['id']
customer_ranking["firstname"] = customers['firstname']
customer_ranking["lastname"] = customers['lastname']

customer_ranking
```

```python
#Tipus de dades del DataFrame 'customer_ranking'
customer_ranking.dtypes
```

```python
#For loop en el DataFrame 'customer_ranking'
for index_customer_ranking, row_product_customer in customer_ranking.iterrows():    
    res = 0
    #For loop en el DataFrame 'df_merged'.
    for index_orders, row_orders in df_merged.iterrows():
        if customer_ranking['id'][index_customer_ranking] == df_merged['customer'][index_orders]:
            res += df_merged['euros'][index_orders]

    #Guardar resultat del total_euros que el client ha gastat en productes.
    customer_ranking.loc[index_customer_ranking, 'total_euros'] = res

#Ordenar les columnes de forma descendent pel camp 'total_euros'.
customer_ranking = customer_ranking.sort_values(by="total_euros", ascending=False)
customer_ranking
```

```python
#Exportar el DataFrame en un fitxer CSV (customer_ranking.csv)
customer_ranking.to_csv("customer_ranking.csv", index=False)
```
